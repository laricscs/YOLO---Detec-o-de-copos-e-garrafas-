```python
import cv2

# Caminho do vídeo original.
caminho_do_video = "video/20261009_115421.mp4"

# Abre o vídeo para leitura.
cap = cv2.VideoCapture(caminho_do_video)

# Verifica se o vídeo foi aberto corretamente.
if not cap.isOpened():
    print("Erro ao carregar vídeo!")

else:
    # Obtém a largura e a altura do vídeo original.
    largura = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    altura = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

    # Obtém a taxa de quadros por segundo (FPS).
    fps = cap.get(cv2.CAP_PROP_FPS)

    # Define o codec para salvar o vídeo em formato MP4.
    fourcc = cv2.VideoWriter_fourcc(*"mp4v")

    # Cria o arquivo de vídeo que receberá as detecções.
    saida = cv2.VideoWriter(
        "video/resultado_deteccao.mp4",
        fourcc,
        fps,
        (largura, altura)
    )

    # Enquanto houver frames para ler...
    while cap.isOpened():

        # Lê o próximo frame do vídeo.
        sucesso, frame = cap.read()

        # Verifica se o frame foi lido corretamente.
        if not sucesso:
            break

        # Executa o YOLOv11 no frame, com confiança mínima de 50%.
        resultados = modelo_v11(frame, conf=0.5, stream=True)

        # Percorre os resultados da detecção.
        for resultado in resultados:

            # Desenha as caixas, classes e confianças no frame.
            frame_anotado = resultado.plot()

            # Exibe o vídeo com as detecções.
            cv2.imshow("Detecção em vídeo", frame_anotado)

            # Grava o frame anotado no vídeo de saída.
            saida.write(frame_anotado)

        # Pressione Q para interromper a execução.
        if cv2.waitKey(1) & 0xFF == ord("q"):
            break

    # Libera os recursos utilizados.
    cap.release()
    saida.release()
    cv2.destroyAllWindows()

    print("Vídeo com detecções salvo em video/resultado_deteccao.mp4")
```
